```python
from datasets import load_dataset

hwu = load_dataset("DeepPavlov/hwu64")

print(hwu)
```

```python
import pandas as pd

hwu_intents = load_dataset(
    "DeepPavlov/hwu64",
    "intents"
)

hwu_intent_df = hwu_intents["intents"].to_pandas()

print(hwu_intent_df[["id", "name"]].head())
```

```python
train_df = hwu["train"].to_pandas()
test_df = hwu["test"].to_pandas()
```

```python
def add_intent_name(df):
    df = df.merge(
        hwu_intent_df[["id", "name"]],
        left_on="label",
        right_on="id",
        how="left"
    )

    df = df.rename(columns={"name": "intent"})
    df = df.drop(columns=["id"])

    return df
```

```python
train_df = add_intent_name(train_df)
test_df = add_intent_name(test_df)
```

```python
from collections import defaultdict

label_examples = defaultdict(list)

for row in hwu["train"]:
    label = row["label"]

    if len(label_examples[label]) < 3:
        label_examples[label].append(row["utterance"])

for label in sorted(label_examples):
    print(f"\nLABEL: {label}")
    for example in label_examples[label]:
        print(" -", example)
```

```python
print("Rows before cleaning:", len(train_df))
print("Intents before cleaning:", train_df["intent"].nunique())
```

```python
REMOVE_INTENTS = {
    "general_affirm",
    "general_commandstop",
    "general_dontcare",
    "general_joke",
    "general_negate",
    "general_praise",
    "general_quirky",
    "general_repeat",
    "general_explain",
}
```

```python
missing_intents = REMOVE_INTENTS - set(train_df["intent"])

print("Removal intents not found:")
print(missing_intents)
```

```python
train_clean = train_df[
    ~train_df["intent"].isin(REMOVE_INTENTS)
].copy()

print("Rows after cleaning:", len(train_clean))
print("Intents after cleaning:", train_clean["intent"].nunique())
```

```python
remaining_removed = set(train_clean["intent"]) & REMOVE_INTENTS

print("Removed intents still present:")
print(remaining_removed)
```

```python
intent_counts = train_clean["intent"].value_counts()

print("Number of intents:", len(intent_counts))
print("Largest class:", intent_counts.max())
print("Smallest class:", intent_counts.min())
print("Average samples per intent:", intent_counts.mean())
```

```python
train_clean.to_csv("../hwu64_train_cleaned.csv", index=False)

print("HWU64 cleaned training dataset saved successfully!")
```

```python
test_clean = test_df[
    ~test_df["intent"].isin(REMOVE_INTENTS)
].copy()
```

```python
print("TEST SET")
print("Rows before:", len(test_df))
print("Rows after:", len(test_clean))
print("Intents after:", test_clean["intent"].nunique())
```

```python
remaining_removed = set(test_clean["intent"]) & REMOVE_INTENTS

print("Removed intents still present:")
print(remaining_removed)
```

```python
test_clean.to_csv("../hwu64_test_cleaned.csv", index=False)

print("HWU64 test dataset saved successfully!")
```
