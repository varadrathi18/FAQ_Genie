```python
from datasets import load_dataset
import pandas as pd

clinc = load_dataset("DeepPavlov/clinc150")

print(clinc)
```

```python
clinc_intents = load_dataset(
    "DeepPavlov/clinc150",
    "intents"
)

intent_df = clinc_intents["intents"].to_pandas()

print(intent_df[["id", "name"]].head())
```

```python
train_df = clinc["train"].to_pandas()
validation_df = clinc["validation"].to_pandas()
test_df = clinc["test"].to_pandas()
```

```python
def add_intent_name(df):
    df = df.merge(
        intent_df[["id", "name"]],
        left_on="label",
        right_on="id",
        how="left"
    )

    df = df.rename(columns={"name": "intent"})
    df = df.drop(columns=["id"])

    return df
```

```python
train_df = add_intent_name(train_df)
validation_df = add_intent_name(validation_df)
test_df = add_intent_name(test_df)
```

```python
train_df["intent"] = train_df["intent"].fillna("out_of_scope")
validation_df["intent"] = validation_df["intent"].fillna("out_of_scope")
test_df["intent"] = test_df["intent"].fillna("out_of_scope")
```

```python
REMOVE_INTENTS = {
    "are_you_a_bot",
    "change_accent",
    "change_ai_name",
    "change_speed",
    "change_user_name",
    "do_you_have_pets",
    "goodbye",
    "greeting",
    "how_old_are_you",
    "meaning_of_life",
    "what_are_your_hobbies",
    "what_can_i_ask_you",
    "what_is_your_name",
    "where_are_you_from",
    "who_do_you_work_for",
    "who_made_you",
    "tell_joke",
    "thank_you",
    "flip_coin",
    "fun_fact",
    "roll_dice",
    "maybe",
    "no",
    "repeat",
    "yes",
    "user_name"
}

print("Intents to remove:", len(REMOVE_INTENTS))
```

```python
print("Rows before cleaning:", len(train_df))
print("Intents before cleaning:", train_df["intent"].nunique())
```

```python
missing_intents = REMOVE_INTENTS - set(train_df["intent"])

print("Removal intents not found:")
print(missing_intents)
```

```python
train_clean = train_df[
    ~train_df["intent"].isin(REMOVE_INTENTS)
].copy()
```

```python
print("Rows after cleaning:", len(train_clean))
print("Intents after cleaning:", train_clean["intent"].nunique())
```

```python
remaining_removed = set(train_clean["intent"]) & REMOVE_INTENTS

print("Removed intents still present:")
print(remaining_removed)
```

```python
intent_counts = train_clean["intent"].value_counts()

print("Number of intents:", len(intent_counts))
print("Largest class:", intent_counts.max())
print("Smallest class:", intent_counts.min())
print("Average samples per intent:", intent_counts.mean())
```

```python
train_clean.to_csv("../clinc150_train_cleaned.csv", index=False)

print("Saved successfully!")
print("File: ../clinc150_train_cleaned.csv")
```

```python
validation_clean = validation_df[
    ~validation_df["intent"].isin(REMOVE_INTENTS)
].copy()

test_clean = test_df[
    ~test_df["intent"].isin(REMOVE_INTENTS)
].copy()
```

```python
print("VALIDATION")
print("Rows before:", len(validation_df))
print("Rows after:", len(validation_clean))
print("Intents after:", validation_clean["intent"].nunique())

print("\nTEST")
print("Rows before:", len(test_df))
print("Rows after:", len(test_clean))
print("Intents after:", test_clean["intent"].nunique())
```

```python
validation_clean.to_csv("../clinc150_validation_cleaned.csv", index=False)
test_clean.to_csv("../clinc150_test_cleaned.csv", index=False)

print("\nCLINC150 validation and test sets saved successfully!")
```
